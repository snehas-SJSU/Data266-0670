````python
# standing parameters — same step 0 as assignment 1
import random
from pathlib import Path

import numpy as np

SID4 = 670
SEED = SID4
SLICE = SID4 % 1000
HP_ID = SID4 % 6
CLS_A = SID4 % 10
CLS_B = (CLS_A + 1 + ((SID4 // 10) % 9)) % 10

random.seed(SEED)
np.random.seed(SEED)
try:
    import torch
    torch.manual_seed(SEED)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)
except ImportError:
    torch = None

print("SID4:", SID4)
print("SEED:", SEED)
print("SLICE:", SLICE)
print("HP_ID:", HP_ID)
print("CLS_A:", CLS_A)
print("CLS_B:", CLS_B)
print("HP_ID is reported only. no extra HP model this hw")
print("torch:", None if torch is None else torch.__version__)
print("cuda:", None if torch is None else torch.cuda.is_available())
ROOT = Path(".").resolve()
for cand in [ROOT, ROOT / "HW-2.5"]:
    if (cand / "scripts" / "run_all.py").exists():
        ROOT = cand
        break
print("HW-2.5 root:", ROOT)
````

````python
# load UUID-labelled JSON after the GPU runs
import json

results = ROOT / "results"
runs = sorted(p for p in results.iterdir() if p.is_dir() and any(p.glob("part_*.json"))) if results.exists() else []
print("result folders:", [p.name for p in runs] or "(none yet — run run_full.bat on the lab GPU)")
for d in runs:
    print("\n==", d.name, "==")
    for name in [
        "part_a_provenance.json",
        "part_b_matmul.json",
        "part_c_roofline.json",
        "part_d_attention.json",
        "part_e_thermal.json",
    ]:
        p = d / name
        if not p.exists():
            print(name, "missing")
            continue
        blob = json.loads(p.read_text())
        gpu = blob.get("gpu") or {}
        print(name, "uuid=", gpu.get("uuid") or blob.get("uuid"))
````
